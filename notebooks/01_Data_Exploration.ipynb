```python

# Create a copy of the original dataset
df_clean = df.copy()

print("Working dataset created successfully!")
print("Original dataset shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python

# Count rows before cleaning
before_cancel = len(df_clean)

# Identify cancelled invoices
cancelled_mask = df_clean["InvoiceNo"].astype(str).str.startswith("C")

# Remove cancelled invoices
df_clean = df_clean.loc[~cancelled_mask].copy()

# Calculate removed rows
removed_cancelled = before_cancel - len(df_clean)

print("CANCELLED INVOICE CLEANING")
print("Rows before:", before_cancel)
print("Cancelled rows removed:", removed_cancelled)
print("Rows after:", len(df_clean))
```

```python

# Count rows before cleaning
before_adjustments = len(df_clean)

# Identify non-cancelled negative quantity adjustments
adjustment_mask = (
    (df_clean["Quantity"] < 0) &
    (df_clean["UnitPrice"] == 0) &
    (df_clean["CustomerID"].isna())
)

# Remove these adjustment records
df_clean = df_clean.loc[~adjustment_mask].copy()

# Calculate removed rows
removed_adjustments = before_adjustments - len(df_clean)

print("ADJUSTMENT RECORD CLEANING")
print("Rows before:", before_adjustments)
print("Adjustment rows removed:", removed_adjustments)
print("Rows after:", len(df_clean))
```

```python

# Count rows before cleaning
before_price = len(df_clean)

# Identify negative-price records
negative_price_mask = df_clean["UnitPrice"] < 0

# Remove negative-price records
df_clean = df_clean.loc[~negative_price_mask].copy()

# Calculate removed rows
removed_negative_prices = before_price - len(df_clean)

print("NEGATIVE PRICE CLEANING")
print("Rows before:", before_price)
print("Negative-price rows removed:", removed_negative_prices)
print("Rows after:", len(df_clean))
```

```python

# Count rows before removing duplicates
before_duplicates = len(df_clean)

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates().copy()

# Calculate removed duplicates
removed_duplicates = before_duplicates - len(df_clean)

print("DUPLICATE CLEANING")
print("Rows before:", before_duplicates)
print("Duplicate rows removed:", removed_duplicates)
print("Rows after:", len(df_clean))
```

```python
print("REMAINING DUPLICATE ROWS:", df_clean.duplicated().sum())
print("CURRENT DATASET SHAPE:", df_clean.shape)

print("\nDUPLICATE COUNT IN ORIGINAL DATA:")
print(df.duplicated().sum())
```

```python

# Count missing descriptions before handling
missing_before = df_clean["Description"].isnull().sum()

# Replace missing descriptions
df_clean["Description"] = df_clean["Description"].fillna("Unknown")

# Count missing descriptions after handling
missing_after = df_clean["Description"].isnull().sum()

print("MISSING DESCRIPTION HANDLING")
print("Missing descriptions before:", missing_before)
print("Missing descriptions after:", missing_after)
print("Current dataset shape:", df_clean.shape)
```

```python

# Identify zero-price transactions
zero_price_df = df_clean[df_clean["UnitPrice"] == 0]

print("ZERO PRICE TRANSACTION ANALYSIS")
print("Total zero-price rows:", len(zero_price_df))

print("\nProduct descriptions:")
display(
    zero_price_df["Description"]
    .value_counts()
    .head(15)
)

print("\nSample zero-price transactions:")
display(
    zero_price_df[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "CustomerID", "Country"]
    ].head(15)
)
```

```python

# Create a flag for zero-price transactions
df_clean["IsZeroPrice"] = df_clean["UnitPrice"] == 0

print("ZERO-PRICE FLAG CREATED")
print("Zero-price transactions:", df_clean["IsZeroPrice"].sum())
print("Normal-price transactions:", (~df_clean["IsZeroPrice"]).sum())

print("\nUpdated dataset shape:", df_clean.shape)

# Display a few flagged records
display(
    df_clean.loc[
        df_clean["IsZeroPrice"],
        ["InvoiceNo", "Description", "Quantity",
         "UnitPrice", "IsZeroPrice"]
    ].head()
)
```

```python

# Identify transactions with missing CustomerIDs
df_clean["IsCustomerIDMissing"] = df_clean["CustomerID"].isna()

print("CUSTOMER ID HANDLING")
print("Missing CustomerIDs:", df_clean["IsCustomerIDMissing"].sum())
print("Available CustomerIDs:", df_clean["CustomerID"].notna().sum())

print("\nUpdated dataset shape:", df_clean.shape)

# Display sample records with missing CustomerIDs
display(
    df_clean.loc[
        df_clean["IsCustomerIDMissing"],
        ["InvoiceNo", "Description", "CustomerID",
         "IsCustomerIDMissing"]
    ].head()
)
```

```python

print("FINAL DATA QUALITY VALIDATION")

print("\n1. Dataset shape:")
print(df_clean.shape)

print("\n2. Remaining missing values:")
print(df_clean.isnull().sum())

print("\n3. Remaining duplicate rows:")
print(df_clean.duplicated().sum())

print("\n4. Negative quantities:")
print((df_clean["Quantity"] < 0).sum())

print("\n5. Negative prices:")
print((df_clean["UnitPrice"] < 0).sum())

print("\n6. Cancelled invoices:")
print(
    df_clean["InvoiceNo"].astype(str).str.startswith("C").sum()
)

print("\n7. Date data type:")
print(df_clean["InvoiceDate"].dtype)

print("\n8. Zero-price transactions:")
print(df_clean["IsZeroPrice"].sum())

print("\n9. Missing CustomerID flags:")
print(df_clean["IsCustomerIDMissing"].sum())
```

```python
# Create transaction-level features

df_clean["TotalAmount"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

df_clean["InvoiceMonth"] = df_clean["InvoiceDate"].dt.month

df_clean["InvoiceYear"] = df_clean["InvoiceDate"].dt.year

df_clean["InvoiceDay"] = df_clean["InvoiceDate"].dt.day

print("Transaction features created successfully!")

df_clean[
    ["InvoiceNo", "Quantity", "UnitPrice", "TotalAmount",
     "InvoiceMonth", "InvoiceYear", "InvoiceDay"]
].head()
```

```python
# 1. Count unique orders for each customer
customer_order_counts = (
    df_clean.dropna(subset=["CustomerID"])
    .groupby("CustomerID")["InvoiceNo"]
    .nunique()
)

df_clean["CustomerTotalOrders"] = df_clean["CustomerID"].map(
    customer_order_counts
)


# 2. Calculate each customer's average order value
customer_invoice_totals = (
    df_clean.dropna(subset=["CustomerID"])
    .groupby(["CustomerID", "InvoiceNo"])["TotalAmount"]
    .sum()
)

customer_avg_order_value = (
    customer_invoice_totals.groupby("CustomerID").mean()
)

df_clean["AverageOrderValue"] = df_clean["CustomerID"].map(
    customer_avg_order_value
)


# 3. Calculate total spending for each customer
customer_spending = (
    df_clean.dropna(subset=["CustomerID"])
    .groupby("CustomerID")["TotalAmount"]
    .sum()
)

df_clean["TotalCustomerSpending"] = df_clean["CustomerID"].map(
    customer_spending
)


print("Customer features created successfully!")

print("\nDataset shape:")
print(df_clean.shape)

print("\nCustomer feature preview:")
print(
    df_clean[
        [
            "CustomerID",
            "CustomerTotalOrders",
            "AverageOrderValue",
            "TotalCustomerSpending"
        ]
    ].head(10)
)
```

```python
print("FEATURE ENGINEERING VALIDATION")

# 1. Check all new columns
feature_columns = [
    "TotalAmount",
    "InvoiceMonth",
    "InvoiceYear",
    "InvoiceDay",
    "CustomerTotalOrders",
    "AverageOrderValue",
    "TotalCustomerSpending"
]

print("\n1. New feature columns:")
print(df_clean[feature_columns].head())

# 2. Verify TotalAmount calculation
print("\n2. TotalAmount calculation errors:")
print(
    (df_clean["TotalAmount"] !=
     df_clean["Quantity"] * df_clean["UnitPrice"]).sum()
)

# 3. Check missing customer IDs
print("\n3. Missing CustomerID rows:")
print(df_clean["CustomerID"].isna().sum())

print("\n4. Missing customer-level feature values:")
print(
    df_clean[
        [
            "CustomerTotalOrders",
            "AverageOrderValue",
            "TotalCustomerSpending"
        ]
    ].isna().sum()
)

# 5. Check customer feature values
print("\n5. Customer feature statistics:")
print(
    df_clean[
        [
            "CustomerTotalOrders",
            "AverageOrderValue",
            "TotalCustomerSpending"
        ]
    ].describe()
)
```
